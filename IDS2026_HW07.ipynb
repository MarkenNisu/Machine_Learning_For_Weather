**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
df["time"] = pd.to_datetime(
    df["time"].str.replace(r"\s+", " ", regex=True),
    format="%Y-%m-%d %H:%M:%S",
)
```

```python
models = ["ECMWF", "GFS", "ICON_EU"]

for m in models:
    df[f"err_{m}"] = df[f"temperature_{m}"] - df["actual_temperature"]

bins = np.arange(np.floor(df["actual_temperature"].min()),
                 np.ceil(df["actual_temperature"].max()) + 2, 2)
df["t_bin"] = pd.cut(df["actual_temperature"], bins)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
for ax, m in zip(axes, models):
    ax.scatter(df["actual_temperature"], df[f"err_{m}"],
               s=4, alpha=0.15, color="gray")

    g = df.groupby("t_bin", observed=True)
    x = g["actual_temperature"].mean()
    ax.plot(x, g[f"err_{m}"].mean(), color="red", lw=2, label="mean error")
    ax.fill_between(x, g[f"err_{m}"].quantile(0.1),
                    g[f"err_{m}"].quantile(0.9),
                    color="red", alpha=0.15, label="10-90%")

    ax.axhline(0, color="black", lw=1)
    ax.set_title(m)
    ax.set_xlabel("Actual temperature (°C)")
axes[0].set_ylabel("Forecast − actual (°C)")
axes[0].legend()
plt.tight_layout()
plt.show()
```

```python

features = {
    "Temperature":   ("actual_temperature",   "temperature",   "°C"),
    "Precipitation": ("actual_precipitation", "precipitation", "mm"),
    "Wind speed":    ("actual_wind_speed",    "wind_speed",    "m/s"),
}

rows = []
for feat, (act, prefix, unit) in features.items():
    for m in models:
        rows.append(pd.DataFrame({
            "feature": feat,
            "model": m,
            "error": df[f"{prefix}_{m}"] - df[act],
            "actual": df[act],
        }))
long = pd.concat(rows, ignore_index=True)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, (feat, (_, _, unit)) in zip(axes, features.items()):
    d = long[long["feature"] == feat].dropna(subset=["error"])
    # choose only "wet" hours
    if feat == "Precipitation":
        d = d[d["actual"] > 0.1]
    sns.violinplot(data=d, x="model", y="error", ax=ax,
                   inner="quartile", cut=0)
    ax.axhline(0, color="black", lw=1)
    ax.set_title(feat + (" (wet hours)" if feat == "Precipitation" else ""))
    ax.set_ylabel(f"Forecast − actual ({unit})")
    ax.set_xlabel("")
plt.tight_layout()
plt.show()
```

**Answer:** Some notes about the features:
- Temperature seems to be predicted quite well by ECMWF and ICON_EU but the errors change when temperature changes. In higher temperatures they underpredict and in lower temperatures they overpredict. The GFS model however seems to be underpredicting in the whole temperature range. This can be seen from the second chart as well, where ECMWF and ICON_EU have a median at or very close to the actual value while GFS has a median of approximately 0.7 degrees lower than actual. It is also worth noting that the GFS model has the biggest spread of the three for predicting temperature.
- No clear differences in models with predicting precipitation. All have medians below zero, which means that it generally rains more than the models predict. It also looks like all the models have trouble with predicting extreme rain, missing by 20 mm per hour or more.
- ICON_EU models seems to be predicting wind speed the best with the smallest spread and median closest to the real value. GFS and ECMWF constantly overpredict wind speed. This could be a problem of the exact location of the cell (see notes in question b), since some locations can be more sheltered than others even at quite small distances.
